```python
#Activity 2 practice
# Download PDF
import urllib.request
# Direct PDF URL
url = "https://arxiv.org/pdf/2307.09288.pdf"


```

```python
print(url)
```

```python
import requests

response = requests.get(url)
response.raise_for_status()

with open("rag_vs_finetuning_paper.pdf", "wb") as f:
    f.write(response.content)

print("PDF downloaded successfully!")
```

```python
loader = PyPDFLoader("rag_vs_finetuning_paper.pdf")

pages = loader.load_and_split()

print(pages[0].page_content)
```

```python
#Practice:Download website HTML auto
import requests

url = "https://dm-academy.app"

response = requests.get(url)

with open("dm_academy.html", "w", encoding="utf-8") as f:
    f.write(response.text)

print("HTML saved successfully")
```

```python
from langchain_community.document_loaders import UnstructuredHTMLLoader

loader = UnstructuredHTMLLoader("dm_academy.html")

data = loader.load()

print(data[0].page_content)
```
