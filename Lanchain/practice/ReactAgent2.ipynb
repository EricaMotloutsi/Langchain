```python
#Create Financial Report Function
def financial_report(company_name: str, revenue: float, expenses: float) -> str:
    """
    Generate a financial report for a company.
    Calculates net income using revenue and expenses.
    """
    net_income = revenue - expenses
    return f"Company: {company_name}, Net Income: {net_income}"
```

```python
from langchain.tools import tool

@tool
def financial_report(company_name: str, revenue: float, expenses: float) -> str:
    """
    Generate a financial report for a company.
    Calculates net income using revenue and expenses.
    """
    net_income = revenue - expenses
    return f"Company: {company_name}, Net Income: {net_income}"
```

```python
print(financial_report.name)
print(financial_report.description)
print(financial_report.args)
```

```python
# Integrate Tool with Agent

import os
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI

load_dotenv()

openai_api_key = os.getenv("OPENAI_API_KEY")

llm = ChatOpenAI(
    model="gpt-4o-mini",
    api_key=openai_api_key,
    temperature=0.1,
    max_completion_tokens=100
)

agent = create_agent(
    llm,
    tools=[financial_report]
)
```

```python
result = agent.invoke({
    "messages": [
        "Create a financial report for Tesla with revenue 100000 and expenses 70000"
    ]
})

print(result["messages"][-1].content)
```
